Visualizing missing rows, and remedy

```python
#print missing rows in California dataset
print("Rows with missing values in California dataset:")
print(df_cal[df_cal.isnull().any(axis=1)])
print("Rows with missing values in Global dataset:")
print(df_global[df_global.isnull().any(axis=1)]) 

#creating new dataframe after dropping missing values in California dataset
df_cal_cleaned = df_cal.dropna()
#confirming that there are no missing values in the new dataframe
print("Missing values in cleaned California dataset:")
print(df_cal_cleaned.isnull().sum())
```

The rows with missing values in the california dataset have missing `AVG_WIND_SPEED` and `WIND_TEMP_RATIO` data. Dropping rows can result in data loss but only 12 entries would be missing so it is fine to remove just those.

```python
# Dropped date column as it is not important and redundant, Also had lots of entries that was empty
df_cal_clean = df_cal.drop(columns = ['DATE'])
# Dropped rows with null values as only 12 enteries was missing compared to 14976 entires that are complete
df_cal_clean = df_cal_clean.dropna()
# Changed Fire Start Day column from boolean to int for easier use, 0: false and 1: true
df_cal_clean['FIRE_START_DAY'] = df_cal_clean['FIRE_START_DAY'].astype(int)
# Change Season column from string to int for easier use, 1: Winter, 2: Spring, 3: Summer, 4: Fall
df_cal_clean['SEASON'] = df_cal_clean['SEASON'].map({
    'Winter': 1,
    'Spring': 2,
    'Summer': 3,
    'Fall': 4
})
print(df_cal_clean)
df_cal_clean.isnull().sum()
```

Some changes to have the datasets be more similar and easier to compare.

```python
# Lower case the titles
df_cal_clean.columns = df_cal_clean.columns.str.lower()
display(df_cal_clean.head(10))

# Renamed occured in df_global_clean to fire_start_day to match df_cal_clean
df_global_clean = df_global.rename(columns={'occured': 'fire_start_day'})

# Change float datatype of daynight_N and fire_start_day to int as the values are binary making float unnecessary
df_global_clean['fire_start_day'] = df_global_clean['fire_start_day'].astype(int)
df_global_clean['daynight_N'] = df_global_clean['daynight_N'].astype(int)
df_global_clean.head(10)
```
