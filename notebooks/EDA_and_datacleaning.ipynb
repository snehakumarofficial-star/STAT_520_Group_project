# Cleaning of the two datasets

Importing packages and datasets

```python
#importing required libraries 

import pandas as pd
import numpy as np

#importing california dataset
df_cal = pd.read_csv('/Users/snehakumar/Documents/Work/Data projects/Project 1/CA_Weather_Fire_Dataset_1984-2025 (1).csv')

#importing global dataset
df_global = pd.read_csv('/Users/snehakumar/Documents/Work/Data projects/Project 1/final_dataset.csv')
```

Inspecting the shape, datatypes, missing values and duplicates in both datasets

```python
print("Shape of California dataset:", df_cal.shape)
print("Shape of Global dataset:", df_global.shape)

print(df_cal.head())
print(df_global.head())

print("Datatypes in California dataset:")
print(df_cal.dtypes)
print("Datatypes in Global dataset:")
print(df_global.dtypes) 


print("Missing values in California dataset:")
print(df_cal.isnull().sum()) 
print("Missing values in Global dataset:")
print(df_global.isnull().sum()) 

print("Duplicate rows in California dataset:", df_cal.duplicated().sum())
print("Duplicate rows in Global dataset:", df_global.duplicated().sum())
```

Results:
* CA dataset: 14988 rows and 14 columns
* Global dataset: 118858 rows and 17 columns

Datatype notes:
* CA dataset: `Date`, Global dataset: all variables are float

Missing values:
* CA dataset: `PRECIPITATION` = 1, `MAX_TEMP` = 1, `MIN_TEMP` = 1, `AVG_WIND_SPEED` = 12, `TEMP_RANGE` = 1, `WIND_TEMP_RATIO`= 12
* Global dataset: no missing values

Duplicate rows: none in both datasets



Visualizing missing rows, and remedy

```python
#print missing rows in California dataset
print("Rows with missing values in California dataset:")
print(df_cal[df_cal.isnull().any(axis=1)])
print("Rows with missing values in Global dataset:")
print(df_global[df_global.isnull().any(axis=1)]) 

#creating new dataframe after dropping missing values in California dataset
df_cal_cleaned = df_cal.dropna()
#confirming that there are no missing values in the new dataframe
print("Missing values in cleaned California dataset:")
print(df_cal_cleaned.isnull().sum())
```

The rows with missing values in the california dataset have missing `AVG_WIND_SPEED` and `WIND_TEMP_RATIO` data. Dropping rows can result in data loss and hence, a new cleaned dataset `df_cal_cleaned` has been created to be used in analysis not using wind or related metrics.

```python
#cleaning up column names in California dataset, changing to lowercase
print("Column names in California dataset before cleaning:")
print(df_cal_cleaned.columns)
df_cal_cleaned.columns = df_cal_cleaned.columns.str.lower() 
print("Column names in California dataset after cleaning:")
print(df_cal_cleaned.columns)
```
